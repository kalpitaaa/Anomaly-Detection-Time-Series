# Question 1: What is Anomaly Detection? Explain its types (point, contextual, andcollective anomalies) with examples.

-->  **Anomaly Detection**

**Anomaly Detection** is the process of identifying data points, events, or patterns that **deviate significantly from normal behavior** in a dataset. These unusual patterns are called **anomalies** or **outliers** and may indicate important events such as fraud, faults, errors, or rare but meaningful occurrences.

Anomaly detection is widely used in areas like fraud detection, network security, medical diagnosis, industrial monitoring, and data quality control.

---

## **Types of Anomalies**

### **1. Point Anomalies**

A **point anomaly** is a **single data instance** that is significantly different from the rest of the data.

**Explanation:**
One data point stands out on its own as abnormal.

**Example:**

* In a dataset of daily temperatures ranging from **20°C to 30°C**, a recorded temperature of **60°C** is a point anomaly.
* In banking, a **$10,000 transaction** on an account that usually spends **$50–$100** is a point anomaly.

---

### **2. Contextual Anomalies**

A **contextual anomaly** is a data point that is **normal in one context but abnormal in another**.
It depends on **contextual attributes** such as time, location, or season.

**Explanation:**
The data value is anomalous only when considering its surrounding context.

**Example:**

* A temperature of **10°C**:

  * Normal in **winter**
  * Anomalous in **summer**
* High website traffic at **2 AM** may be anomalous, but the same traffic at **8 PM** may be normal.

---

### **3. Collective Anomalies**

A **collective anomaly** occurs when a **group of related data points together** is anomalous, even if individual points may appear normal.

**Explanation:**
The anomaly is detected only when considering the sequence or collection as a whole.

**Example:**

* In network traffic, a sudden sequence of small data packets may indicate a **DDoS attack**, even though each packet looks normal.
* In ECG (heart monitoring) data, a sequence of normal-looking signals together may indicate a **heart rhythm disorder**.


# Question 2: Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of their approach and suitable use cases.

-->  **Comparison of Isolation Forest, DBSCAN, and Local Outlier Factor (LOF)**

Isolation Forest, DBSCAN, and Local Outlier Factor are popular **unsupervised anomaly detection techniques**, but they differ in how they identify anomalies and where they work best.

---

## **1. Isolation Forest (IF)**

### **Approach**

* Works by **isolating data points** instead of modeling normal behavior.
* Uses random feature selection and random split values to build **isolation trees**.
* Anomalies are isolated **faster** (with fewer splits) because they are rare and different.

### **Key Idea**

> Anomalies are data points that are easier to isolate.

### **Suitable Use Cases**

* Large, **high-dimensional datasets**
* Fraud detection
* Network intrusion detection
* Situations where anomalies are **few and different**

### **Advantages**

* Scales well to large datasets
* Handles high-dimensional data
* No distance or density calculation

### **Limitations**

* Less interpretable
* May struggle with strongly clustered anomalies

---

## **2. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)**

### **Approach**

* A **density-based clustering** algorithm.
* Groups points into clusters based on density.
* Points that do not belong to any dense cluster are labeled as **outliers (noise)**.

### **Key Idea**

> Normal points belong to dense regions; anomalies lie in sparse regions.

### **Suitable Use Cases**

* Datasets with **spatial or geographic structure**
* Arbitrary-shaped clusters
* Noise detection in clustering problems

### **Advantages**

* Does not require the number of clusters in advance
* Finds clusters of any shape
* Naturally detects noise as anomalies

### **Limitations**

* Sensitive to parameter selection (`eps`, `minPts`)
* Struggles with **varying densities**
* Less effective in high-dimensional data

---

## **3. Local Outlier Factor (LOF)**

### **Approach**

* A **density-based anomaly detection** method.
* Compares the **local density** of a point with that of its neighbors.
* Points with significantly lower density than neighbors are anomalies.

### **Key Idea**

> Anomalies have much lower local density than their neighbors.

### **Suitable Use Cases**

* Detecting **local anomalies**
* Data with varying density
* Fraud detection where anomalies occur near normal points

### **Advantages**

* Detects **contextual/local outliers**
* Works well when global methods fail

### **Limitations**

* Computationally expensive for large datasets
* Sensitive to choice of neighbors (`k`)
* Performance drops in very high dimensions

# Question 3: What are the key components of a Time Series? Explain each with one example

--> **Key Components of a Time Series**

A **time series** is a sequence of data points collected or recorded at **regular time intervals** (e.g., hourly, daily, monthly). Time series analysis aims to understand the underlying structure of the data over time.

A time series is commonly decomposed into the following **four key components**:

---

## **1. Trend (T)**

### **Explanation**

The **trend** represents the **long-term movement** or general direction of the data over time. It shows whether the data is increasing, decreasing, or remaining stable.

### **Example**

* The **annual sales of electric vehicles** increasing steadily over several years show an **upward trend**.

---

## **2. Seasonal Component (S)**

### **Explanation**

The **seasonal component** captures **regular, repeating patterns** that occur at fixed intervals (daily, monthly, yearly).

### **Example**

* **Ice cream sales** rising every summer and falling every winter demonstrate **seasonality**.

---

## **3. Cyclical Component (C)**

### **Explanation**

The **cyclical component** represents **long-term fluctuations** that occur over irregular periods, often influenced by economic or external factors.

### **Example**

* **Economic growth and recession cycles** affecting stock market performance over several years illustrate cyclical behavior.

---

## **4. Irregular / Random Component (I)**

### **Explanation**

The **irregular component** consists of **unpredictable, random variations** that cannot be attributed to trend, seasonality, or cycles.

### **Example**

* A sudden drop in airline bookings due to an **unexpected natural disaster** is an irregular component.


# Question 4: Define Stationary in time series. How can you test and transform a non-stationary series into a stationary one?

-->  **Stationarity in Time Series**

#### **Definition**

A time series is said to be **stationary** if its **statistical properties remain constant over time**. Specifically:

* **Mean** is constant
* **Variance** is constant
* **Autocovariance / autocorrelation** depends only on the **lag**, not on time

Stationarity is important because most time series models (like **ARMA/ARIMA**) assume the data is stationary.

---

## **Testing for Stationarity**

### **1. Visual Inspection**

* **Time series plot**: look for trends or changing variance.
* **Rolling statistics**: plot rolling mean and rolling variance; if they change over time, the series is likely non-stationary.

### **2. Augmented Dickey–Fuller (ADF) Test**

* **Null hypothesis (H₀):** The series has a unit root (non-stationary).
* If **p-value < 0.05**, reject H₀ → series is **stationary**.

### **3. KPSS Test**

* **Null hypothesis (H₀):** The series is stationary.
* If **p-value < 0.05**, reject H₀ → series is **non-stationary**.

> Using both ADF and KPSS together gives a more reliable conclusion.

---

## **Transforming a Non-Stationary Series into a Stationary One**

### **1. Differencing**

* Subtract the previous value from the current value:
  [
  y'*t = y_t - y*{t-1}
  ]
* Removes **trend**.
* Seasonal differencing removes **seasonality**.

**Example:**
Daily stock prices → first difference (price changes)

---

### **2. Log / Power Transformation**

* Apply **log**, **square root**, or **Box–Cox** transformations.
* Stabilizes **changing variance** (heteroscedasticity).

**Example:**
Log-transforming sales data with rapidly increasing variability

---

### **3. Detrending**

* Remove the trend using regression or moving averages.
* Residuals are often stationary.

**Example:**
Subtracting a fitted linear trend from GDP data

---

### **4. Seasonal Adjustment**

* Remove seasonal patterns using decomposition.
* Common in monthly or quarterly data.

**Example:**
Removing holiday effects from retail sales



# Question 5: Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in terms of structure and application.

--> ### **Differences Between AR, MA, ARIMA, SARIMA, and SARIMAX Models**

These models are widely used for **time series forecasting**. They differ in structure, complexity, and the type of data they are best suited for.

---

## **1. AR (Autoregressive) Model**

### **Structure**

* Uses past values of the series to predict the current value.
* Represented as **AR(p)**, where *p* is the number of lagged observations.

[
y_t = c + \phi_1 y_{t-1} + \phi_2 y_{t-2} + \dots + \phi_p y_{t-p} + \varepsilon_t
]

### **Application**

* When the series is **stationary** and depends on its own past values.
* Example: Forecasting **daily temperature** using previous days’ temperatures.

---

## **2. MA (Moving Average) Model**

### **Structure**

* Uses past **error terms** (shocks) to predict the current value.
* Represented as **MA(q)**, where *q* is the number of lagged error terms.

[
y_t = c + \varepsilon_t + \theta_1 \varepsilon_{t-1} + \dots + \theta_q \varepsilon_{t-q}
]

### **Application**

* Suitable when random shocks affect the series.
* Example: Modeling **noise-like financial returns**.

---

## **3. ARIMA (Autoregressive Integrated Moving Average)**

### **Structure**

* Combines AR and MA models with **differencing** to handle non-stationarity.
* Represented as **ARIMA(p, d, q)**:

  * *p*: AR order
  * *d*: differencing order
  * *q*: MA order

### **Application**

* For **non-stationary time series** without seasonality.
* Example: Forecasting **monthly sales** with a trend.

---

## **4. SARIMA (Seasonal ARIMA)**

### **Structure**

* Extends ARIMA by adding **seasonal components**.
* Represented as:

[
SARIMA(p, d, q)(P, D, Q)_s
]

where *s* is the seasonal period (e.g., 12 for monthly data).

### **Application**

* For time series with **trend and seasonality**.
* Example: Forecasting **monthly airline passenger data**.

---

## **5. SARIMAX (Seasonal ARIMA with eXogenous variables)**

### **Structure**

* Further extends SARIMA by including **external (exogenous) variables**.
* Includes all SARIMA terms plus regression on external factors.

### **Application**

* When external variables influence the time series.
* Example: Predicting **electricity demand** using temperature and humidity.

---

## **Comparison Table**

| Model   | Structure                 | Handles Trend | Handles Seasonality | Uses External Variables | Typical Use Case            |
| ------- | ------------------------- | ------------- | ------------------- | ----------------------- | --------------------------- |
| AR      | AR(p)                     | ❌             | ❌                   | ❌                       | Stationary series           |
| MA      | MA(q)                     | ❌             | ❌                   | ❌                       | Shock-driven series         |
| ARIMA   | AR + I + MA               | ✅             | ❌                   | ❌                       | Non-stationary data         |
| SARIMA  | ARIMA + seasonal terms    | ✅             | ✅                   | ❌                       | Seasonal data               |
| SARIMAX | SARIMA + exogenous inputs | ✅             | ✅                   | ✅                       | External-factor-driven data |

---

### **Conclusion**

* Use **AR/MA** for **stationary** data.
* Use **ARIMA** for **trended** data.
* Use **SARIMA** for **seasonal** data.
* Use **SARIMAX** when **external factors** affect the series.

```python
#Question 7: Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to detect anomalies. Visualize the anomalies on a 2D scatter plot.
#(Include your Python code and output in the code box below.)

# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Set random seed for reproducibility
np.random.seed(42)

# Create a synthetic NYC Taxi Fare-like dataset
n_samples = 500

trip_distance = np.random.normal(loc=5, scale=2, size=n_samples)
fare_amount = trip_distance * 3 + np.random.normal(0, 2, n_samples)

# Add anomalies (very high fares and distances)
anomaly_distance = np.random.uniform(20, 40, 20)
anomaly_fare = anomaly_distance * 5 + np.random.uniform(20, 50, 20)

# Combine normal data and anomalies
distance = np.concatenate([trip_distance, anomaly_distance])
fare = np.concatenate([fare_amount, anomaly_fare])

df = pd.DataFrame({
    "trip_distance": distance,
    "fare_amount": fare
})

# Apply Isolation Forest
iso_forest = IsolationForest(
    n_estimators=100,
    contamination=0.04,
    random_state=42
)

df["anomaly"] = iso_forest.fit_predict(df[["trip_distance", "fare_amount"]])

# Convert labels: -1 = anomaly, 1 = normal
df["anomaly"] = df["anomaly"].map({1: 0, -1: 1})

# Separate normal points and anomalies
normal = df[df["anomaly"] == 0]
anomalies = df[df["anomaly"] == 1]

# Plot results
plt.figure(figsize=(8, 6))
plt.scatter(normal["trip_distance"], normal["fare_amount"],
            c="blue", label="Normal", alpha=0.6)
plt.scatter(anomalies["trip_distance"], anomalies["fare_amount"],
            c="red", label="Anomaly", edgecolors="black")

plt.xlabel("Trip Distance (miles)")
plt.ylabel("Fare Amount ($)")
plt.title("Isolation Forest Anomaly Detection on NYC Taxi Fare Data")
plt.legend()
plt.show()
```

```python
#Question 9: Apply Local Outlier Factor (LOF) on any numerical dataset to detect anomalies and visualize them using matplotlib.
#(Include your Python code and output in the code box below.)

# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# Set random seed for reproducibility
np.random.seed(42)

# Create a synthetic numerical dataset
# Normal data
X_normal = 0.3 * np.random.randn(300, 2)

# Add clustered data
X_cluster = np.random.randn(200, 2) + np.array([3, 3])

# Add anomalies (outliers)
X_outliers = np.random.uniform(low=-6, high=6, size=(20, 2))

# Combine all data
X = np.vstack([X_normal, X_cluster, X_outliers])

# Apply Local Outlier Factor
lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.05
)

y_pred = lof.fit_predict(X)

# Convert labels: -1 = anomaly, 1 = normal
anomalies = X[y_pred == -1]
normal = X[y_pred == 1]

# Plot results
plt.figure(figsize=(8, 6))
plt.scatter(normal[:, 0], normal[:, 1],
            c="blue", label="Normal", alpha=0.6)
plt.scatter(anomalies[:, 0], anomalies[:, 1],
            c="red", label="Anomaly", edgecolors="black")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Local Outlier Factor (LOF) Anomaly Detection")
plt.legend()
plt.show()
```

```python
#Question 10: You are working as a data scientist for a power grid monitoring company.
#Your goal is to forecast energy demand and also detect abnormal spikes or drops in
#real-time consumption data collected every 15 minutes. The dataset includes features
#like timestamp, region, weather conditions, and energy usage.
#Explain your real-time data science workflow:
#● How would you detect anomalies in this streaming data (Isolation Forest / LOF /DBSCAN)?
#● Which time series model would you use for short-term forecasting (ARIMA / SARIMA / SARIMAX)?
#● How would you validate and monitor the performance over time?
#● How would this solution help business decisions or operations?


## **Real-Time Data Science Workflow**

### **1. Anomaly Detection in Streaming Data**

#* **Model:** **Isolation Forest** (primary)
#* **Why:** Works in high-dimensional data, fast for streaming, detects global anomalies.
#* **Features:** Energy usage, lagged usage, weather conditions, region.
#* **Implementation:** Sliding window training (e.g., last 7 days), score each new 15-min data point, flag anomalies.
#* Optional: **LOF** for local anomalies, DBSCAN for offline cluster-based anomaly analysis.


### **2. Short-Term Forecasting**

#* **Model:** **SARIMAX**
#* **Why:** Handles trend, daily/weekly seasonality, and exogenous variables like weather or region.
#* Forecast horizon: Next 1–24 hours (or next 1–96 intervals at 15-min resolution).



### **3. Validation & Monitoring**

#* **Forecast:** MAE, RMSE, MAPE over rolling windows.
#* **Anomalies:** Track false positives/negatives; recalibrate thresholds.
#* **Monitoring:** Detect data drift; retrain Isolation Forest daily, SARIMAX weekly or as performance drops.



### **4. Business Impact**

#* Early detection of abnormal spikes → prevent outages or equipment stress.
#* Accurate short-term forecasts → optimize generator scheduling, load balancing.
#* Reduce energy costs and improve grid reliability.


## **Python Example (Synthetic Streaming Data)**


# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Simulate streaming-like 15-min energy usage data
np.random.seed(42)
timestamps = pd.date_range(start='2025-01-01', periods=96*7, freq='15T')  # 1 week
temperature = 20 + 5*np.sin(np.linspace(0, 14*np.pi, len(timestamps))) + np.random.randn(len(timestamps))
energy_usage = 50 + 10*np.sin(np.linspace(0, 14*np.pi, len(timestamps))) + 0.5*temperature + np.random.randn(len(timestamps))
region = ['A']*len(timestamps)

df = pd.DataFrame({
    'timestamp': timestamps,
    'region': region,
    'temperature': temperature,
    'energy_usage': energy_usage
})
df.set_index('timestamp', inplace=True)

# -------------------------------
# 1. Anomaly Detection (Isolation Forest)
# -------------------------------
iso_forest = IsolationForest(contamination=0.02, random_state=42)
df['anomaly'] = iso_forest.fit_predict(df[['energy_usage','temperature']])
df['anomaly'] = df['anomaly'].map({1:0, -1:1})

# Plot anomalies
plt.figure(figsize=(12,4))
plt.plot(df.index, df['energy_usage'], label='Energy Usage')
plt.scatter(df.index[df['anomaly']==1], df['energy_usage'][df['anomaly']==1],
            color='red', label='Anomaly')
plt.xlabel('Time')
plt.ylabel('Energy Usage (kWh)')
plt.title('Isolation Forest Anomaly Detection (Streaming Data)')
plt.legend()
plt.show()

# -------------------------------
# 2. Short-Term Forecasting (SARIMAX)
# -------------------------------
# Aggregate to hourly for faster SARIMAX fitting
df_hourly = df.resample('H').mean()

model = SARIMAX(df_hourly['energy_usage'],
                order=(1,1,1),
                seasonal_order=(1,1,1,24),
                exog=df_hourly[['temperature']],
                enforce_stationarity=False,
                enforce_invertibility=False)
results = model.fit(disp=False)

# Forecast next 24 hours
forecast = results.get_forecast(steps=24, exog=df_hourly[['temperature']].iloc[-24:])
forecast_mean = forecast.predicted_mean
forecast_ci = forecast.conf_int()

# Plot forecast
plt.figure(figsize=(12,4))
plt.plot(df_hourly.index, df_hourly['energy_usage'], label='Observed')
plt.plot(forecast_mean.index, forecast_mean, color='red', label='Forecast')
plt.fill_between(forecast_ci.index,
                 forecast_ci.iloc[:,0],
                 forecast_ci.iloc[:,1],
                 color='pink', alpha=0.3)
plt.xlabel('Time')
plt.ylabel('Energy Usage (kWh)')
plt.title('SARIMAX Short-Term Forecast')
plt.legend()
plt.show()

### **Output Explanation**

#1. **Anomaly Detection Plot:**

   #* Red points → abnormal spikes or drops in energy usage.
   #* Blue line → normal energy consumption.

#2. **Forecast Plot:**

   #* Red line → predicted energy usage for next 24 hours
  # * Pink shaded region → 95% confidence interval


```
